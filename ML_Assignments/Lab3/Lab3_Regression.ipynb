# Lab 3 - Các Kỹ thuật Hồi quy (Regression Techniques)

Yêu cầu:
1. **Giảm Gradient (Gradient Descent), Hồi quy tuyến tính (Linear Regression):** Code thuật toán Gradient Descent từ con số 0 (Scratch). So sánh tốc độ hội tụ với các learning rate khác nhau.
2. **Hồi quy đa thức (Polynomial Regression), Hồi quy logistic (Logistic Regression):** Dự đoán khả năng rời bỏ dịch vụ (Churn) hoặc Phân loại hoa Iris dùng Logistic Regression.

## 1. Thuật toán Giảm Gradient (Gradient Descent) từ con số 0 (Scratch)

```python
import numpy as np
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt

# Tạo dữ liệu ngẫu nhiên có tính tuyến tính
np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
X_b = np.c_[np.ones((100, 1)), X]  # Thêm x0 = 1 vào mỗi điểm dữ liệu

def plot_gradient_descent(theta, eta, theta_path=None):
    m = len(X_b)
    plt.plot(X, y, "b.")
    n_iterations = 1000
    X_new = np.array([[0], [2]])
    X_new_b = np.c_[np.ones((2, 1)), X_new]
    
    for iteration in range(n_iterations):
        if iteration < 10:
            y_predict = X_new_b.dot(theta)
            style = "b-" if iteration > 0 else "r--"
            plt.plot(X_new, y_predict, style)
        gradients = 2/m * X_b.T.dot(X_b.dot(theta) - y)
        theta = theta - eta * gradients
        if theta_path is not None:
            theta_path.append(theta)
    plt.xlabel("$x_1$", fontsize=18)
    plt.axis([0, 2, 0, 15])
    plt.title(r"$\eta = {}$".format(eta), fontsize=16)

np.random.seed(42)
theta = np.random.randn(2,1)  # Khởi tạo ngẫu nhiên

plt.figure(figsize=(15,4))
plt.subplot(141); plot_gradient_descent(theta, eta=0.001)
plt.ylabel("$y$", rotation=0, fontsize=18)
plt.subplot(142); plot_gradient_descent(theta, eta=0.01)
plt.subplot(143); plot_gradient_descent(theta, eta=0.1)
plt.subplot(144); plot_gradient_descent(theta, eta=1.0)

plt.show()
```

## 2. Hồi quy đa thức (Polynomial Regression)

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

m = 100
X = 6 * np.random.rand(m, 1) - 3
y = 0.5 * X**2 + X + 2 + np.random.randn(m, 1)

poly_features = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly_features.fit_transform(X)

lin_reg = LinearRegression()
lin_reg.fit(X_poly, y)

X_new = np.linspace(-3, 3, 100).reshape(100, 1)
X_new_poly = poly_features.transform(X_new)
y_new = lin_reg.predict(X_new_poly)

plt.plot(X, y, "b.")
plt.plot(X_new, y_new, "r-", linewidth=2, label="Dự đoán (Predictions)")
plt.xlabel("$x_1$", fontsize=18)
plt.ylabel("$y$", rotation=0, fontsize=18)
plt.legend(loc="upper left", fontsize=14)
plt.axis([-3, 3, 0, 10])
plt.title("Hồi quy đa thức (Polynomial Regression)")
plt.show()
```

## 3. Hồi quy Logistic (Logistic Regression) - Phân loại hoa Iris

```python
from sklearn import datasets
from sklearn.linear_model import LogisticRegression

iris = datasets.load_iris()
X = iris["data"][:, 3:]  # Chiều rộng cánh hoa (petal width)
y = (iris["target"] == 2).astype(int)  # Bằng 1 nếu là Iris virginica, ngược lại bằng 0

log_reg = LogisticRegression(solver="lbfgs", random_state=42)
log_reg.fit(X, y)

X_new = np.linspace(0, 3, 1000).reshape(-1, 1)
y_proba = log_reg.predict_proba(X_new)
decision_boundary = X_new[y_proba[:, 1] >= 0.5][0]

plt.figure(figsize=(8, 3))
plt.plot(X[y==0], y[y==0], "bs")
plt.plot(X[y==1], y[y==1], "g^")
plt.plot([decision_boundary, decision_boundary], [-1, 2], "k:", linewidth=2)
plt.plot(X_new, y_proba[:, 1], "g-", linewidth=2, label="Iris virginica")
plt.plot(X_new, y_proba[:, 0], "b--", linewidth=2, label="Không phải Iris virginica")
plt.text(decision_boundary+0.02, 0.15, "Biên giới quyết định\n(Decision Boundary)", fontsize=14, color="k", ha="center")
plt.xlabel("Chiều rộng cánh hoa (cm)", fontsize=14)
plt.ylabel("Xác suất", fontsize=14)
plt.legend(loc="center left", fontsize=14)
plt.axis([0, 3, -0.02, 1.02])
plt.title("Logistic Regression (Nhận diện hoa Iris-Virginica)")
plt.show()
```
